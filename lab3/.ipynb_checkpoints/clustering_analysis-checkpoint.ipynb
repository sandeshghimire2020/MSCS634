## Step 1: Load and Prepare the Dataset

The Wine dataset holds thirteen chemical measurements from 178 Italian wines grown by three cultivars. Clustering ignores those cultivar labels during fitting and tries to recover the group structure from the chemistry alone; the labels are kept aside purely to score the result afterwards.

Both algorithms here partition by distance, so feature scale decides what "near" means. The lab brief calls for z-score standardisation, and the final section of this notebook shows concretely why that instruction matters.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')

from sklearn.datasets import load_wine
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score, silhouette_samples, adjusted_rand_score
from scipy.spatial.distance import cdist

plt.rcParams['figure.figsize'] = (10, 6)
plt.rcParams['axes.grid'] = True

wine = load_wine()
X_raw = wine.data
y_true = wine.target

df = pd.DataFrame(X_raw, columns=wine.feature_names)
df['cultivar'] = [wine.target_names[i] for i in y_true]

print("Dataset Shape:", X_raw.shape)
print("Features:", len(wine.feature_names))
print("True classes:", len(wine.target_names))
print("\nFirst Five Rows:")
print(df.head())
```

```python
# Class distribution
print("Class Distribution:")
print("=" * 55)
counts = df['cultivar'].value_counts().sort_index()
for name, n in counts.items():
    print(f"  {name:10} {n:4d} wines  ({n/len(df)*100:5.1f}%)")
print(f"\nTotal: {len(df)} wines")

print("\nThese labels are NOT used to fit either model. They serve only to compute the")
print("Adjusted Rand Index afterwards, which asks how closely the discovered partition")
print("matches the real cultivars. The classes are mildly uneven, which matters because")
print("both algorithms tested here have a mild preference for similarly-sized clusters.")
```

```python
# Feature scales before standardisation
print("Feature Scales Before Standardisation:")
print("=" * 72)
print(f"{'feature':32} {'min':>10} {'max':>10} {'std':>10}")
print("-" * 72)
for i, name in enumerate(wine.feature_names):
    col = X_raw[:, i]
    print(f"{name:32} {col.min():10.2f} {col.max():10.2f} {col.std():10.2f}")

var_share = X_raw.var(axis=0) / X_raw.var(axis=0).sum() * 100
dominant = np.argmax(var_share)
print(f"\n'{wine.feature_names[dominant]}' alone carries {var_share[dominant]:.2f}% of total variance.")
print("Left unscaled, Euclidean distance would be almost entirely a measurement of that")
print("one chemical, and the remaining twelve features would barely influence any cluster.")
```

```python
# Z-score standardisation: every feature gets mean 0 and standard deviation 1
scaler = StandardScaler()
X = scaler.fit_transform(X_raw)

print("After Z-Score Standardisation:")
print("=" * 55)
summary = pd.DataFrame({
    'mean': X.mean(axis=0),
    'std': X.std(axis=0),
    'min': X.min(axis=0),
    'max': X.max(axis=0)
}, index=wine.feature_names)
print(summary.round(4))

print(f"\nAll means are now 0 (max deviation {np.abs(X.mean(axis=0)).max():.2e}) and all")
print("standard deviations are 1, so every chemical contributes to distance on equal terms.")
```

---
## Step 2: Implement K-Means Clustering

K-Means represents each cluster by its **centroid**, the arithmetic mean of its members. It alternates between assigning points to the nearest centroid and recomputing those centroids, minimising the total squared distance from points to their assigned centre.

The centroid is a constructed average, not an observation. It will generally not correspond to any wine that actually exists.

`k = 3` is chosen to match the three known cultivars, and `n_init=10` restarts the algorithm from ten different random initialisations and keeps the best, which guards against a poor starting position.

```python
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
kmeans_labels = kmeans.fit_predict(X)

kmeans_sil = silhouette_score(X, kmeans_labels)
kmeans_ari = adjusted_rand_score(y_true, kmeans_labels)

print("K-Means Results (k = 3)")
print("=" * 55)
print(f"  Silhouette Score      : {kmeans_sil:.4f}")
print(f"  Adjusted Rand Index   : {kmeans_ari:.4f}")
print(f"  Inertia (within-SS)   : {kmeans.inertia_:.2f}")
print(f"  Iterations to converge: {kmeans.n_iter_}")
print(f"\n  Cluster sizes: {np.bincount(kmeans_labels)}")
print(f"  True class sizes: {np.bincount(y_true)}")

print("\nReading the two metrics:")
print("  Silhouette runs from -1 to +1 and uses no labels. It asks whether each point sits")
print("  closer to its own cluster than to the next nearest one, so it measures geometric")
print("  tightness and separation.")
print("  ARI runs from 0 (chance) to 1 (identical) and compares the partition against the")
print("  true cultivars, correcting for agreements that random chance would produce.")
```

```python
# Where do the clusters fall relative to the real cultivars?
print("K-Means Clusters vs. True Cultivars:")
print("=" * 55)
ct = pd.crosstab(pd.Series([wine.target_names[i] for i in y_true], name='true cultivar'),
                 pd.Series(kmeans_labels, name='k-means cluster'))
print(ct)

errors = len(y_true) - ct.max(axis=1).sum()
print(f"\nUnder the best cluster-to-class matching, {errors} of {len(y_true)} wines are misgrouped.")
print("Two cultivars are recovered exactly; essentially all the confusion sits in the third.")
```

---
## Step 3: Implement K-Medoids Clustering

K-Medoids swaps the centroid for a **medoid**: an actual data point chosen so the total distance from it to the rest of its cluster is as small as possible. Because the centre must be a real observation, it cannot be dragged off into empty space by an extreme value, which makes the method more robust to outliers than K-Means.

### A note on implementation

`sklearn` has no K-Medoids. The usual third-party source, `scikit-learn-extra`, was last released in 2022 and does not work against the `sklearn` version installed here, so rather than pin an abandoned dependency the Partitioning Around Medoids (PAM) algorithm is implemented directly below. PAM has two phases:

1. **BUILD** — greedily pick the k starting medoids, taking first the point closest to all others, then repeatedly whichever remaining point most reduces total cost.
2. **SWAP** — consider exchanging every medoid with every non-medoid, keep any swap that lowers total cost, and repeat until no swap helps.

This is the standard formulation and uses only `numpy` and `scipy`.

```python
def k_medoids_pam(X, k, max_iter=300):
    """Partitioning Around Medoids. Returns (medoid_indices, labels, total_cost, passes)."""
    D = cdist(X, X)              # full pairwise distance matrix
    n = len(X)

    # --- BUILD phase: greedy initial medoid selection ---
    medoids = [int(np.argmin(D.sum(axis=1)))]
    while len(medoids) < k:
        nearest = D[medoids].min(axis=0)
        best, best_cost = None, np.inf
        for cand in range(n):
            if cand in medoids:
                continue
            cost = np.minimum(nearest, D[cand]).sum()
            if cost < best_cost:
                best_cost, best = cost, cand
        medoids.append(best)
    medoids = np.array(sorted(medoids))

    # --- SWAP phase: exchange medoids with non-medoids while cost falls ---
    cost = D[medoids].min(axis=0).sum()
    passes = 0
    for passes in range(1, max_iter + 1):
        improved = False
        for slot in range(k):
            for cand in range(n):
                if cand in medoids:
                    continue
                trial = medoids.copy()
                trial[slot] = cand
                trial_cost = D[trial].min(axis=0).sum()
                if trial_cost < cost - 1e-12:
                    medoids, cost, improved = trial, trial_cost, True
        if not improved:
            break

    labels = np.argmin(D[medoids], axis=0)
    return medoids, labels, cost, passes

print("PAM implementation defined.")
```

```python
medoid_indices, medoid_labels, medoid_cost, passes = k_medoids_pam(X, k=3)

medoid_sil = silhouette_score(X, medoid_labels)
medoid_ari = adjusted_rand_score(y_true, medoid_labels)

print("K-Medoids Results (k = 3)")
print("=" * 55)
print(f"  Silhouette Score    : {medoid_sil:.4f}")
print(f"  Adjusted Rand Index : {medoid_ari:.4f}")
print(f"  Total cost (sum of distances to medoid): {medoid_cost:.2f}")
print(f"  Converged after {passes} swap pass(es)")
print(f"\n  Cluster sizes: {np.bincount(medoid_labels)}")
print(f"\n  Medoid row indices: {medoid_indices}")
print("\nThe medoids are real wines. Their actual chemistry:")
print(pd.DataFrame(X_raw[medoid_indices], columns=wine.feature_names,
                   index=[f'medoid_{i}' for i in range(3)]).round(2).T)
```

```python
print("K-Medoids Clusters vs. True Cultivars:")
print("=" * 55)
ct_med = pd.crosstab(pd.Series([wine.target_names[i] for i in y_true], name='true cultivar'),
                     pd.Series(medoid_labels, name='k-medoids cluster'))
print(ct_med)

errors_med = len(y_true) - ct_med.max(axis=1).sum()
print(f"\nUnder the best matching, {errors_med} of {len(y_true)} wines are misgrouped,")
print(f"compared with {errors} for K-Means.")
print(f"\nAgreement between the two partitions: ARI = {adjusted_rand_score(kmeans_labels, medoid_labels):.4f}")
print("They broadly agree, so the two methods are finding the same structure and differing")
print("mainly over where to place the boundary.")
```

## Step 4: Visualize and Compare Results

The data has thirteen dimensions, so a scatter plot needs a projection. Principal Component Analysis provides one by rotating the data onto the axes of greatest variance.

This projection is lossy, and the loss is reported below. Points that appear to overlap on the page may be well separated along a discarded dimension, so the plot should be read as a reasonable sketch rather than a faithful map.

```python
pca = PCA(n_components=2, random_state=42)
X_pca = pca.fit_transform(X)

kmeans_centres_pca = pca.transform(kmeans.cluster_centers_)
medoid_points_pca = X_pca[medoid_indices]

ev = pca.explained_variance_ratio_
print(f"PC1 explains {ev[0]*100:.1f}% of variance, PC2 explains {ev[1]*100:.1f}%.")
print(f"The two-dimensional view below retains {ev.sum()*100:.1f}% of the total variance,")
print(f"which means roughly {(1-ev.sum())*100:.0f}% of the structure is not visible in it.")
```

```python
fig, axes = plt.subplots(1, 2, figsize=(16, 6.5))
palette = ['#45B7D1', '#FF6B6B', '#5BC98C']

# --- K-Means ---
for c in range(3):
    m = kmeans_labels == c
    axes[0].scatter(X_pca[m, 0], X_pca[m, 1], s=55, alpha=0.65,
                    color=palette[c], label=f'cluster {c} (n={m.sum()})',
                    edgecolors='white', linewidths=0.5)
axes[0].scatter(kmeans_centres_pca[:, 0], kmeans_centres_pca[:, 1],
                marker='X', s=420, c='black', edgecolors='yellow',
                linewidths=2.2, label='centroid', zorder=10)
axes[0].set_title(f'K-Means  |  silhouette {kmeans_sil:.3f}  |  ARI {kmeans_ari:.3f}',
                  fontsize=13, fontweight='bold')

# --- K-Medoids ---
for c in range(3):
    m = medoid_labels == c
    axes[1].scatter(X_pca[m, 0], X_pca[m, 1], s=55, alpha=0.65,
                    color=palette[c], label=f'cluster {c} (n={m.sum()})',
                    edgecolors='white', linewidths=0.5)
axes[1].scatter(medoid_points_pca[:, 0], medoid_points_pca[:, 1],
                marker='D', s=330, c='black', edgecolors='yellow',
                linewidths=2.2, label='medoid (a real wine)', zorder=10)
axes[1].set_title(f'K-Medoids  |  silhouette {medoid_sil:.3f}  |  ARI {medoid_ari:.3f}',
                  fontsize=13, fontweight='bold')

for ax in axes:
    ax.set_xlabel(f'PC1 ({ev[0]*100:.1f}% of variance)', fontsize=11, fontweight='bold')
    ax.set_ylabel(f'PC2 ({ev[1]*100:.1f}% of variance)', fontsize=11, fontweight='bold')
    ax.legend(fontsize=9, loc='best')

plt.suptitle('Cluster Assignments Projected onto Two Principal Components',
             fontsize=14, fontweight='bold', y=1.00)
plt.tight_layout()
plt.show()
```

```python
# True cultivars shown on the same projection, for reference
fig, ax = plt.subplots(figsize=(9, 6.5))
for c, name in enumerate(wine.target_names):
    m = y_true == c
    ax.scatter(X_pca[m, 0], X_pca[m, 1], s=55, alpha=0.65, color=palette[c],
               label=f'{name} (n={m.sum()})', edgecolors='white', linewidths=0.5)
ax.set_xlabel(f'PC1 ({ev[0]*100:.1f}%)', fontsize=11, fontweight='bold')
ax.set_ylabel(f'PC2 ({ev[1]*100:.1f}%)', fontsize=11, fontweight='bold')
ax.set_title('Ground Truth: Actual Cultivars', fontsize=13, fontweight='bold')
ax.legend(fontsize=10)
plt.tight_layout()
plt.show()

print("Comparing this against the two panels above shows how faithfully each algorithm")
print("recovered the real groups without ever being shown them.")
```

```python
# Metric comparison
comparison = pd.DataFrame({
    'Silhouette Score': [kmeans_sil, medoid_sil],
    'Adjusted Rand Index': [kmeans_ari, medoid_ari],
    'Misgrouped wines': [errors, errors_med],
    'Cluster sizes': [str(np.bincount(kmeans_labels)), str(np.bincount(medoid_labels))]
}, index=['K-Means', 'K-Medoids'])

print("Side-by-Side Comparison")
print("=" * 78)
print(comparison.to_string())

print(f"\nK-Means leads on silhouette by {kmeans_sil - medoid_sil:+.4f}")
print(f"K-Means leads on ARI by        {kmeans_ari - medoid_ari:+.4f}")

fig, axes = plt.subplots(1, 2, figsize=(13, 5))
for ax, col, ttl in zip(axes, ['Silhouette Score', 'Adjusted Rand Index'],
                        ['Silhouette Score (higher = tighter clusters)',
                         'Adjusted Rand Index (higher = closer to truth)']):
    bars = ax.bar(comparison.index, comparison[col], color=['#45B7D1', '#FF6B6B'], width=0.55)
    ax.set_title(ttl, fontsize=12, fontweight='bold')
    ax.set_ylim(0, max(comparison[col]) * 1.25)
    for b, v in zip(bars, comparison[col]):
        ax.text(b.get_x() + b.get_width()/2, v + max(comparison[col])*0.03,
                f'{v:.4f}', ha='center', fontsize=11, fontweight='bold')
plt.tight_layout()
plt.show()
```

### Analysis

**Which algorithm produced better-defined clusters?** K-Means, on both metrics. Its silhouette score is slightly higher, and its ARI is substantially higher. The gap in ARI is the meaningful one: K-Means misgroups only a handful of wines, while K-Medoids misgroups noticeably more, and the two metrics agree on the ranking rather than pulling in opposite directions.

**What differs in cluster shape and positioning?** Both algorithms find the same broad three-group structure — their partitions agree closely — and both recover two of the three cultivars essentially perfectly. The disagreement is concentrated almost entirely in the third cultivar, the one sitting between the other two in the projection. K-Medoids pulls a block of those middle wines into a neighbouring cluster, producing a larger, more lopsided cluster where K-Means divides the region more evenly.

The reason is where each centre is allowed to sit. A K-Means centroid is an average and can settle anywhere, including in the empty space between groups, which is precisely where a boundary-defining centre is most useful. A K-Medoids medoid must be an actual wine, so it is pinned to a populated location. In a dense region with no clean gap, that constraint costs accuracy.

**Both silhouette scores are low in absolute terms** — under 0.30, where 1.0 would be ideal — even though K-Means achieves an ARI near 0.90. The two metrics are measuring different things. Silhouette asks whether the clusters are geometrically compact and well separated; ARI asks whether they match reality. Here the cultivars genuinely abut one another in chemical space rather than forming isolated balls, so a near-perfect recovery of the true groups still scores modestly on geometry. A low silhouette is not by itself evidence of a bad clustering.

**When is each preferable?**

*K-Means* suits this data and most data like it: numeric features, roughly convex groups, no severe outliers. It is far faster, since PAM must evaluate every medoid-by-non-medoid swap, and its freely-placed centroids position boundaries better in dense regions.

*K-Medoids* earns its cost in three situations. With **outliers**, a centroid gets dragged toward an extreme value while a medoid cannot move off a real observation. With **non-numeric or non-Euclidean data** — categorical features, edit distance, correlation distance — a mean may not even be definable, whereas a medoid only requires a pairwise distance matrix. And when the centre must be **interpretable or actionable**, a medoid is a genuine example you can point to, which matters when the result is "this specific wine typifies the group" rather than an abstract average.

```python
# Why the brief insists on standardisation: the same clustering, scaled vs unscaled
print("Effect of Standardisation on K-Means (k = 3)")
print("=" * 68)
print(f"{'features':>12} {'silhouette':>13} {'ARI':>10}")
print("-" * 68)

for name, data in [('unscaled', X_raw), ('standardised', X)]:
    km = KMeans(n_clusters=3, random_state=42, n_init=10).fit(data)
    print(f"{name:>12} {silhouette_score(data, km.labels_):13.4f} "
          f"{adjusted_rand_score(y_true, km.labels_):10.4f}")

print("\nThis is the most important result in the notebook, and it is a trap.")
print("The unscaled clustering scores MUCH HIGHER on silhouette yet MUCH LOWER on ARI.")
print("\nUnscaled, distance is dominated by the single large-magnitude feature identified in")
print("Step 1, so K-Means slices the data into tight, crisply separated bands along that one")
print("axis. Geometrically those bands look excellent. They simply are not the cultivars.")
print("\nThe practical warning: silhouette uses no ground truth, so on a real problem where")
print("labels are unavailable it would have steered us confidently toward the wrong answer.")
print("An internal metric rewards clusters that are tidy, not clusters that are correct.")
```

```python
# Does silhouette at least pick the right k, once the data is standardised?
print("Choosing k by Silhouette (standardised data)")
print("=" * 55)
print(f"{'k':>3} {'silhouette':>13} {'ARI':>10}")
print("-" * 55)

ks = range(2, 9)
sils, aris = [], []
for k in ks:
    km = KMeans(n_clusters=k, random_state=42, n_init=10).fit(X)
    s = silhouette_score(X, km.labels_)
    a = adjusted_rand_score(y_true, km.labels_)
    sils.append(s); aris.append(a)
    print(f"{k:3d} {s:13.4f} {a:10.4f}{'   <- best silhouette' if s == max(sils) and k > 2 else ''}")

best_k = list(ks)[int(np.argmax(sils))]
print(f"\nSilhouette peaks at k = {best_k}, which matches the three true cultivars.")
print("On standardised data the internal metric does find the right number of groups,")
print("so the failure in the previous cell came from the scaling, not from the metric itself.")

fig, ax = plt.subplots(figsize=(10, 5.5))
ax.plot(list(ks), sils, marker='o', linewidth=2.2, markersize=9,
        color='#45B7D1', label='Silhouette (no labels used)')
ax.plot(list(ks), aris, marker='s', linewidth=2.2, markersize=9,
        color='#FF6B6B', label='ARI (vs true cultivars)')
ax.axvline(3, color='gray', linestyle='--', linewidth=1.4, label='k = 3 (true classes)')
ax.set_xlabel('k', fontsize=12, fontweight='bold')
ax.set_ylabel('Score', fontsize=12, fontweight='bold')
ax.set_title('Both Metrics Peak at the True Number of Clusters', fontsize=13, fontweight='bold')
ax.set_xticks(list(ks))
ax.legend(fontsize=10)
plt.tight_layout()
plt.show()
```

## Conclusion

K-Means and K-Medoids were both fitted with k = 3 to the z-score standardised Wine dataset and scored on silhouette and Adjusted Rand Index.

K-Means won on both counts, and decisively on ARI: it recovered the three cultivars almost exactly while K-Medoids misgrouped roughly two and a half times as many wines. The two methods agreed on the overall structure, and both recovered two cultivars perfectly; their disagreement was confined to the third, which sits between the others in chemical space. The cause is the medoid constraint itself. A centroid is free to settle in the sparse gap between groups, which is where a dividing centre does the most good, whereas a medoid must occupy the position of a real wine. That restriction buys robustness to outliers, and this dataset has none worth worrying about, so here it only costs accuracy.

Two findings extend past the brief. First, both silhouette scores are low — below 0.30 — even where ARI approaches 0.90, because the cultivars genuinely adjoin one another rather than forming separated balls. A modest silhouette score does not imply a poor clustering; the two metrics answer different questions. Second, and more striking: clustering the *unscaled* data yields a far better silhouette score (about 0.57 against 0.28) but a far worse ARI (about 0.37 against 0.90). Unscaled, distance collapses onto the one feature measured in the hundreds, and K-Means carves out tidy bands along that axis which are geometrically crisp and factually wrong. Since silhouette consults no labels, on a genuine unlabelled problem it would have endorsed that result with confidence.

The lesson is that internal validation metrics reward tidiness, not truth. They are useful for choosing k — silhouette does correctly pick k = 3 on the standardised data — but they cannot tell you whether the feature space you built was the right one to begin with. That judgement stays with the analyst.
